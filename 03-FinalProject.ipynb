## Background and Prior Work

Fast fashion is a clothing production model where clothing is produced quickly and cheaply, with rapid turnover due to changing trends and high-volume manufacturing. This way of production has completely changed the fashion industry, how we see trends, and how we consume clothes, especially in the past two decades.

On one hand, this model has been praised for making clothing more affordable to people who may not have been able to afford expensive, high quality clothing. However, it also raises concerns surrounding textile lifespan. Garments are increasingly made in ways that shorten the usable life, and this contributes to waste, pollution, and higher resource demand.
Prior research shows that fast fashion production practices directly reduce textile lifespan. According to the Sustainability Directory, fast fashion brands more often use inexpensive, synthetic fibers in their clothing. These fabrics degrade quickly with regular wear and washing, leading the clothes to lose structural integrity after fewer uses.<a name="#ref1"></a>[<sup>1</sup>](#note1) Additionally, the constant release of new clothing styles encourages consumers to discard more clothing that is not necessarily unusable, but outdated. According to the United States Environmental Protection Agency, the total mass of textile waste generated each year has increased roughly tenfold since 1960 while the total United States population has barely doubled.<a name="#ref4"></a>[<sup>4</sup>](#note4) The Geneva Environment Network also reported that approximately 85% of textiles are discarded each year, mostly ending up in landfills or incinerators rather than being recycled.<a name="#ref2"></a>[<sup>2</sup>](#note2) They found that the fashion industry operates a model where clothing is produced at high volume with low costs, but is rarely kept in circulation for very long.

The environmental effects of fast fashion are commonly discussed, but specific and reliable information regarding numbers is often inconsistent or difficult to prove. A Vox investigation “Fashion has a misinformation problem. That’s bad for the environment.” by Alden Wicker shows that, while there is clear evidence of pollution and overproduction, data on textile waste and garment lifespan are mainly sourced from limited regional studies.<a name="#ref3"></a>[<sup>3</sup>](#note3) They noted that claims about how quickly clothing is thrown out and how much ends up in landfills varies widely across sources. This makes it difficult to regulate the industry at all or evaluate sustainability claims from most companies.

The lack of definite consumer-level data surrounding the garment industry led us to want to examine the clothing use and disposal practices of specific populations, in our case college students. Rather than exclusively using estimated data that can be inconsistent, we will analyze self-reported behaviors from a smaller population. Our sources have provided us with a background for examining how college students consume and ultimately dispose of clothing.

### Tidying

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd

df = pd.read_csv('data/00-raw/Clothing Purchasing Behavior 2.csv')

column_names = ['timestamp', 'consent', 'major', 'new_vs_secondhand', 'purchase_freq', 'avg_price', 'disposal_method', 'disposal_reason']
df.columns = column_names

# Number of rows
df.shape[0]
# Multiple variables are embedded in the `disposal_reason` column, separated by commas.
# What are the unique reasons for disposal?

unique_reasons = set[str]()
for response in df['disposal_reason']:
    unique_reasons.update(response.split(', '))

def count_occurrences(reason: str) -> int:
    return df['disposal_reason'].str.contains(reason).sum()

disposal_reason_df = pd.DataFrame(unique_reasons, columns=['disposal_reason'])
disposal_reason_df['count'] = disposal_reason_df['disposal_reason'].apply(count_occurrences)
disposal_reason_df.sort_values(by='count', ascending=False)
# Give common disposal reasons their own columns; collect singletons in a column called 'other'

df_tidy = df.copy()

disposal_col_names = {
    "Doesn't fit anymore": "disposal_reason_doesnt_fit",
    "Don't like it anymore": "disposal_reason_dont_like",
    "Freeing up space": "disposal_reason_space",
    "Torn/Broken": "disposal_reason_torn",
    "Selling for money": "disposal_reason_money"
}

for reason, new_col in disposal_col_names.items():
    df_tidy[new_col] = df_tidy['disposal_reason'].str.contains(reason).astype(bool)

def is_other(response: str) -> bool:
    reasons = response.split(', ')
    other = False
    for reason in reasons:
        if reason not in [*disposal_col_names.keys()]:
            other = True
    return other

df_tidy['disposal_reason_other'] = df_tidy['disposal_reason'].apply(is_other)
df_tidy = df_tidy.drop(columns=['disposal_reason'])

# Now that each variable has its own column, the dataset is tidy.
df_tidy.to_csv('data/01-interim/survey_tidy2.csv', index=False)
df_tidy.head()
```

#### Major Proportions: Does our survey sample reflect the general UCSD population by major?

Our goal with this EDA is to explore whether our sample is representative of the general UCSD population in terms of major, in the case that major may be  confounding varibale.

We considered that certain majors may be more likely to shop secondhand, such as environmental studies or fashion. However, we want to first see how "good" our data matches UCSD's distribution.

We will do this with a Chi-Square Goodness of Fit Test.

- Null Hypothesis: The survey sample proportion by major does NOT significantly differ from the expected proportions (from the UCSD degrees-conferred dataset). Hidden assumption: The UCSD degrees-conferred dataset is approximately equal to the number of people in a major
- Alternate Hypothesis: The survey sample proportion by major DOES significantly differ from the expected proportions

```python
# Keyword-Based Classifier Function: Maps Major keywords to 7 Categories
# This code block was generated with GPT-4 and edited to reflect common major keywords of UCSD students
def categorize_major(input_major):
    
    if any(word in input_major for word in [
        'literature', 'history', 'philosophy', 'art', 'music',
        'theatre', 'visual', 'foreign language', 'linguistics',
        'humanities', 'arts', 'theater', 'dance', 'english'
    ]):
        return 'humanities' # Arts & Humanities
    
    elif any(word in input_major for word in [
        'bio', 'neuro', 'chem', 'biological',
        'human biology', 'life science'
    ]):
        return 'bio' # Biological Sciences
   
    elif any(word in input_major for word in [
        'health', 'public health'
    ]):
        return 'health' # Health Sciences
    
    elif any(word in input_major for word in [
        'engineering', 'computer', 'electrical',
        'mechanical', 'structural', 'nano', 'aerospace', 'civil', 'industrial'
    ]):
        return 'engineering' # Engineering
    
    elif any(word in input_major for word in [
        'physical', 'physics', 'math', 'mathematics', 'statistics',
        'astronomy', 'physical science'
    ]):
        return 'physical' # Physical Sciences
    
    elif any(word in input_major for word in [
        'social', 'cognitive', 'psychology', 'econ', 'economics', 'sociology',
        'political', 'international', 'anthropology',
        'communication', 'business', 'finance'
    ]):
        return 'social' # Social Sciences
    
    else:
        return 'other'
```

```python
# Apply categorization method to df
df['category'] = df['major'].apply(categorize_major)
```

```python
observed_counts = df['category'].value_counts()

# Reorder/force in case one category doesnt appear (generated by GPT-4)
categories = [
    'humanities',
    'bio',
    'health',
    'engineering',
    'physical',
    'social',
    'other'
]

observed_counts = (
    df['category']
    .value_counts()
    .reindex(categories, fill_value=0)
)

print(observed_counts)
```

```python
from scipy.stats import chi2_contingency

# Create contingency table (counts)
ct = df.groupby(['new_vs_secondhand', 'avg_price']).size().unstack('avg_price').fillna(0)

chi2, p, dof, expected = chi2_contingency(ct)

print("Chi-square Statistic:", chi2)
print("P-value:", p)
```

Now, let's see if the relationship between shopper type and disposal method is statistically significant using a Chi-Square test.

```python
# Is the relationship between shopper type and disposal method statistically significant?
from scipy.stats import chi2_contingency

# Contingency table of raw counts for purchase type vs disposal method
contingency_table = pd.crosstab(df['new_vs_secondhand'], df['disposal_method'])

# Run the Chi-Square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Print the results
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
```

The p-value is 0.23, so the relationship between shopper type and disposal method is not statistically signifcant.

First, some remapping that will be reflected in the clean up code later:

```python
reason_cols = [
    "disposal_reason_doesnt_fit",
    "disposal_reason_dont_like",
    "disposal_reason_space",
    "disposal_reason_torn",
    "disposal_reason_money",
    "disposal_reason_other"
]
df_long = (
    df_clean
    .melt(
        id_vars=["disposal_method"],
        value_vars=reason_cols,
        var_name="disposal_reason",
        value_name="selected"
    )
)
df_long = df_long[df_long["selected"]]
reason_label_map = {
    "disposal_reason_doesnt_fit": "Doesn't fit",
    "disposal_reason_dont_like": "Don't like it",
    "disposal_reason_space": "Freeing space",
    "disposal_reason_torn": "Torn/Broken",
    "disposal_reason_money": "Selling for money",
    "disposal_reason_other": "Other"
}

df_long["disposal_reason"] = df_long["disposal_reason"].map(reason_label_map)
```

```python
def simplify_method(method):
    method = str(method).lower()
    if "donate" in method or "thrift" in method or "hand down" in method:
        return "Donate"
    elif "sell" in method or "ebay" in method or "yard sale" in method:
        return "Sell"
    elif (
        "recycl" in method
        or "cut" in method
        or "scrap" in method
        or "patch" in method
        or "sewing" in method
        or "rag" in method
        or "cleaning" in method
        or "repurpose" in method
    ):
        return "Recycle"
    elif "trash" in method or "throw" in method or "landfill" in method:
        return "Throw Away"
    elif "keep" in method or "closet" in method:
        return "Keep"
    else:
        return "Other"

df_long["disposal_method_clean"] = df_long["disposal_method"].apply(simplify_method)
ct = pd.crosstab(
    df_long["disposal_reason"],
    df_long["disposal_method_clean"]
)
```

```python
ct_percent = ct.div(ct.sum(axis=1), axis=0)

ct_percent.plot(kind="bar", stacked=True)

plt.ylabel("Proportion")
plt.xlabel("Disposal Reason")
plt.title("Disposal Method Distribution by Reason")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

The stacked bar plot displays the distribution of disposal methods across different disposal reasons. Several clear patterns emerge:
- Donation is the dominant disposal method across nearly all reasons (e.g., “Doesn’t Fit,” “Don’t Like It,” “Freeing Space,” and “Torn/Broken”).
- For items discarded due to “Selling for Money,” selling represents a substantial portion of responses, as expected.
- When clothing is “Torn/Broken,” there is a noticeable increase in “Throw Away” and “Recycle” relative to other reasons, though donation still remains common.
- The “Other” category shows more variation across disposal methods, suggesting heterogeneous responses in that group.

Chi-Square Test to see if this is significant:

```python
chi2, p, dof, expected = chi2_contingency(ct)

print("Chi-square Statistic:", chi2)
print("P-value:", p)
print("Degrees of Freedom:", dof)
```

The p-value is 0.0000299, so the relationship between disposal reason and disposal method is statistically signifcant. This suggests that the way students dispose of clothing is not independent of their stated reason for disposal.

The results show a statistically significant association between disposal reason and disposal method. Donation is the most common method overall, regardless of reason. Throwing away increases for damaged clothing but does not dominate.

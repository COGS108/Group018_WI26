## Research Question

Among college students, are disposal reasons and disposal methods of clothes related to how often they are purchased and whether they are new or secondhand?


## Background and Prior Work

Fast fashion is a clothing production model where clothing is produced quickly and cheaply, with rapid turnover due to changing trends and high-volume manufacturing. This way of production has completely changed the fashion industry, how we see trends, and how we consume clothes, especially in the past two decades.

On one hand, this model has been praised for making clothing more affordable to people who may not have been able to afford expensive, high quality clothing. However, it also raises concerns surrounding textile lifespan. Garments are increasingly made in ways that shorten the usable life, and this contributes to waste, pollution, and higher resource demand.
Prior research shows that fast fashion production practices directly reduce textile lifespan. According to the Sustainability Directory, fast fashion brands more often use inexpensive, synthetic fibers in their clothing. These fabrics degrade quickly with regular wear and washing, leading the clothes to lose structural integrity after fewer uses.<a name="#ref1"></a>[<sup>1</sup>](#note1) Additionally, the constant release of new clothing styles encourages consumers to discard more clothing that is not necessarily unusable, but outdated. According to the United States Environmental Protection Agency, the total mass of textile waste generated each year has increased roughly tenfold since 1960 while the total United States population has barely doubled.<a name="#ref4"></a>[<sup>4</sup>](#note4) The Geneva Environment Network also reported that approximately 85% of textiles are discarded each year, mostly ending up in landfills or incinerators rather than being recycled.<a name="#ref2"></a>[<sup>2</sup>](#note2) They found that the fashion industry operates a model where clothing is produced at high volume with low costs, but is rarely kept in circulation for very long.

The environmental effects of fast fashion are commonly discussed, but specific and reliable information regarding numbers is often inconsistent or difficult to prove. A Vox investigation “Fashion has a misinformation problem. That’s bad for the environment.” by Alden Wicker shows that, while there is clear evidence of pollution and overproduction, data on textile waste and garment lifespan are mainly sourced from limited regional studies.<a name="#ref3"></a>[<sup>3</sup>](#note3) They noted that claims about how quickly clothing is thrown out and how much ends up in landfills varies widely across sources. This makes it difficult to regulate the industry at all or evaluate sustainability claims from most companies.

The lack of definite consumer-level data surrounding the garment industry led us to want to examine the clothing use and disposal practices of specific populations, in our case college students. Rather than exclusively using estimated data that can be inconsistent, we will analyze self-reported behaviors from a smaller population. Our sources have provided us with a background for examining how college students consume and ultimately dispose of clothing.

## Hypothesis


We predict that among college students, higher clothing purchase frequency will correlate with a higher likelihood to throw away unwanted clothing. Additionally, we predict that students who buy secondhand clothing will be more likely to donate unwanted clothing, while less likely to dispose of it due to damage. This is based on the Sustainability Directory report that stated fast fashion brands use materials that are less resistant to normal wear and tear, and tend to have shorter release and production cycles, leading to increased consumption and disposal.

### Tidying

```python
import pandas as pd
df = pd.read_csv('data/00-raw/Clothing Purchasing Behavior.csv')
df.columns = [
    'timestamp',
    'consent',
    'major',
    'new_vs_secondhand',
    'purchase_freq',
    'avg_price',
    'disposal_method',
    'disposal_reason'
]
df_tidy = df.copy()
disposal_col_names = {
    "Doesn't fit anymore": "disposal_reason_doesnt_fit",
    "Don't like it anymore": "disposal_reason_dont_like",
    "Freeing up space": "disposal_reason_space",
    "Torn/Broken": "disposal_reason_torn",
    "Selling for money": "disposal_reason_money"
}
for reason, new_col in disposal_col_names.items():
    df_tidy[new_col] = df_tidy['disposal_reason'].str.contains(reason).astype(bool)
def is_other(response: str) -> bool:
    reasons = response.split(', ')
    for reason in reasons:
        if reason not in disposal_col_names:
            return True
    return False

df_tidy['disposal_reason_other'] = df_tidy['disposal_reason'].apply(is_other)
df_tidy = df_tidy.drop(columns=['disposal_reason'])

df_clean = df_tidy.copy()
df_clean['major'] = df_clean['major'].fillna('Unknown')
df_clean['major'] = df_clean['major'].str.strip().str.lower()
df_clean = df_clean.drop(index=3)
df_clean = df_clean[df_clean['consent'] == "Yes"]
df_clean = df_clean.drop(columns=['consent'])

reason_cols = [
    "disposal_reason_doesnt_fit",
    "disposal_reason_dont_like",
    "disposal_reason_space",
    "disposal_reason_torn",
    "disposal_reason_money",
    "disposal_reason_other"
]
df_long = df_clean.melt(
    id_vars=["disposal_method"],
    value_vars=reason_cols,
    var_name="disposal_reason",
    value_name="selected"
)
df_long = df_long[df_long["selected"]]
reason_label_map = {
    "disposal_reason_doesnt_fit": "Doesn't fit",
    "disposal_reason_dont_like": "Don't like it",
    "disposal_reason_space": "Freeing space",
    "disposal_reason_torn": "Torn/Broken",
    "disposal_reason_money": "Selling for money",
    "disposal_reason_other": "Other"
}
df_long["disposal_reason"] = df_long["disposal_reason"].map(reason_label_map)
def simplify_method(method):
    method = str(method).lower()

    if "donate" in method or "thrift" in method or "hand down" in method:
        return "Donate"
    elif "sell" in method or "ebay" in method or "yard sale" in method:
        return "Sell"
    elif ("recycl" in method or "cut" in method or "scrap" in method or "patch" in method or "sewing" in method or "rag" in method or "cleaning" in method or "repurpose" in method):
        return "Recycle"
    elif "trash" in method or "throw" in method or "landfill" in method:
        return "Throw Away"
    elif "keep" in method or "closet" in method:
        return "Keep"
    else:
        return "Other"
df_long["disposal_method_clean"] = df_long["disposal_method"].apply(simplify_method)

# add disposal_method_clean to df_clean as well 
df_clean["disposal_method_clean"] = df_clean["disposal_method"].apply(simplify_method)
# drop the original disposal_method column since we have the cleaned version now
df_clean = df_clean.drop(columns=['disposal_method'])

df_clean.to_csv('data/02-processed/survey_clean2.csv', index=False)
df_clean.head()
```

#### Major Proportions: Does our survey sample reflect the general UCSD population by major?

Our goal with this EDA is to explore whether our sample is representative of the general UCSD population in terms of major, in the case that major may be  confounding varibale.

We considered that certain majors may be more likely to shop secondhand, such as environmental studies or fashion. However, we want to first see how "good" our data matches UCSD's distribution.

We will do this with a Chi-Square Goodness of Fit Test.

- Null Hypothesis: The survey sample proportion by major does NOT significantly differ from the expected proportions (from the UCSD degrees-conferred dataset). Hidden assumption: The UCSD degrees-conferred dataset is approximately equal to the number of people in a major
- Alternate Hypothesis: The survey sample proportion by major DOES significantly differ from the expected proportions

```python
# Keyword-Based Classifier Function: Maps Major keywords to 7 Categories
# This code block was generated with GPT-4 and edited to reflect common major keywords of UCSD students
def categorize_major(input_major):
    
    if any(word in input_major for word in [
        'literature', 'history', 'philosophy', 'art', 'music',
        'theatre', 'visual', 'foreign language', 'linguistics',
        'humanities', 'arts', 'theater', 'dance', 'english'
    ]):
        return 'humanities' # Arts & Humanities
    
    elif any(word in input_major for word in [
        'bio', 'neuro', 'chem', 'biological',
        'human biology', 'life science'
    ]):
        return 'bio' # Biological Sciences
   
    elif any(word in input_major for word in [
        'health', 'public health'
    ]):
        return 'health' # Health Sciences
    
    elif any(word in input_major for word in [
        'engineering', 'computer', 'electrical',
        'mechanical', 'structural', 'nano', 'aerospace', 'civil', 'industrial'
    ]):
        return 'engineering' # Engineering
    
    elif any(word in input_major for word in [
        'physical', 'physics', 'math', 'mathematics', 'statistics',
        'astronomy', 'physical science'
    ]):
        return 'physical' # Physical Sciences
    
    elif any(word in input_major for word in [
        'social', 'cognitive', 'psychology', 'econ', 'economics', 'sociology',
        'political', 'international', 'anthropology',
        'communication', 'business', 'finance'
    ]):
        return 'social' # Social Sciences
    
    else:
        return 'other'
```

```python
# Apply categorization method to df
df['category'] = df['major'].apply(categorize_major)
```

### Shopper Type vs. Average Price


```python
# Shopper type vs. Average price
new_vs_secondhand_map = {
   'Almost always new': 'New',
   'Mostly new, some secondhand': 'New',
   'About equal new and secondhand': 'Equal New & Secondhand',
   'Mostly secondhand, some new': 'Secondhand',
   'Almost always secondhand': 'Secondhand'
}

df['new_vs_secondhand'] = df['new_vs_secondhand'].map(new_vs_secondhand_map)

# Group by purchase type and disposal method, count  number of responses,  reshape for plotting
plot_df = df.groupby(['new_vs_secondhand', 'avg_price']).size().unstack('avg_price').fillna(0)

# Normalize by shopper type to get proportion of shopper groups that use each disposal method
sub_df = plot_df.div(plot_df.sum(axis=1), axis=0)

ax = sub_df.plot(kind='bar', stacked=True, rot=0,
                 title='Shopper type vs. Average price')

ax.legend(title='average price', loc='center left', bbox_to_anchor=(1.0, 0.5))
ax.set_ylabel("Proportion within group")
ax.set_xlabel("Shopper Type")
```

```python
# Create contingency table (counts)
ct = df.groupby(['new_vs_secondhand', 'avg_price']).size().unstack('avg_price').fillna(0)

chi2, p, dof, expected = chi2_contingency(ct)

print("Chi-square Statistic:", chi2)
print("P-value:", p)
```

The stacked barplot displays the proportion of: average price spent per shirt/top vs shopper type. 

Shoppers who buy new clothes are the highest proportion who spend \$15-35 per shirt/top. 

For shoppers who buy new and second hand equally, there's a pretty even split between \$5-15 and \$15-35 per shirt/top. There's a small amount who spend \$35-75 per shirt/top.

For second hand shoppers, there's another even split between \$5-15 and \$15-35 per shirt/top. But there's a small amount of shoppers spending \$0-5 a shirt/top, and a smaller amount of shoppers spending  \$35-75 per shirt/top.

\$75+ per shirt/top is an outlier of average price, located in the new shopper group.

From greatest to least, the average price spent per shirt/top is: 1. new shoppers, 2. equal new / second hand shoppers, 3. and second hand shoppers.

Since p=0.021 < 0.05, there is a statistically significant association between shopper type and average price.

The stacked barplot displays the proportion of disposal methods within each group of shoppers. It shows that the proportion of secondhand shoppers who donate their clothes is less than that of new shoppers. This contradicts our hypothesis that there is a higher proportion of donaters among secondhand shoppers compared to new shoppers. However, the plot does show a higher proportion of throwing away clothes among new shoppers compared to secondhand shoppers. Secondhand shoppers have the highest proportion of recyclers and sellers of unwanted clothes. As for shoppers who buy equal new and secondhand, they are the group with the highest proportion of donaters. 

Now, let's see if the relationship between shopper type and disposal method is statistically significant using a Chi-Square test.

```python
# Contingency table of raw counts for purchase type vs disposal method
contingency_table = pd.crosstab(df_clean['new_vs_secondhand'], df_clean['disposal_method_clean'])

# Run the Chi-Square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Print the results
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
```

The p-value is 0.08, so, at the 5% significance level, the relationship between shopper type and disposal method is not statistically signifcant.

```python
ct = pd.crosstab(df_long["disposal_reason"], df_long["disposal_method_clean"])

sankey_df = ct.reset_index()
sankey_df.columns = ["reason"] + list(ct.columns)
sankey_df = sankey_df.melt(id_vars="reason", var_name="method", value_name="count")
sankey_df = sankey_df[sankey_df["count"] > 0]

reasons = list(ct.index)
methods = list(ct.columns)
labels = reasons + methods
reason_map = {r: i for i, r in enumerate(reasons)}
method_map = {m: i + len(reasons) for i, m in enumerate(methods)}

source = sankey_df["reason"].map(reason_map)
target = sankey_df["method"].map(method_map)
value = sankey_df["count"]
reason_y = [i / (len(reasons) - 1) if len(reasons) > 1 else 0.5 for i in range(len(reasons))]
method_y = [i / (len(methods) - 1) if len(methods) > 1 else 0.5 for i in range(len(methods))]

fig = go.Figure(data=[go.Sankey(
    arrangement="snap",
    node=dict(label=labels, pad=35, thickness=30, x=[0] * len(reasons) + [1] * len(methods), y=reason_y + method_y),
    link=dict(source=source, target=target, value=value))])

fig.update_layout(title_text="Clothing Disposal Flow: Reason → Method", height=700, font_size=13)
fig.show()
#showing diagram as png since Git previews didn't like showing
#fig.write_image("newplot.png", width=1200, height=700)
```

The Sankey diagram illustrates the flow of responses from disposal reasons to disposal methods, allowing us to visualize how students choose to dispose of clothing depending on why they are getting rid of it. Several patterns are visible:
- Donation is the most common disposal method across nearly all reasons. The largest flows originate from reasons such as “Doesn't fit,” - “Don't like it,” and “Freeing space” and lead to the Donate category. This suggests that when clothing is still usable but unwanted, students most often choose donation rather than discarding it.
- Damaged clothing (Torn/Broken) shows more varied disposal behavior. While donation still occurs, there are noticeably larger flows toward Throw Away and Recycle compared with other reasons. This indicates that when garments are damaged, students are more likely to discard or repurpose them rather than donate them.
- Selling for money strongly corresponds with the “Sell” disposal method, which is expected. However, the flow is relatively small compared with donation flows from other reasons, indicating that selling clothing is a less common disposal pathway overall.
- Minor flows appear toward “Keep,” “Recycle,” and “Other” methods, but these represent smaller proportions of responses and do not dominate any particular disposal reason.

### Purchase Frequency vs. Disposal Method

Let's visualize the relationship between reported purchase frequency and reported clothing disposal method. 

```python
# order purchase frequency for better visualization
freq_order = [
    'Every few years',
    'About once a year',
    'Every few months',
    'About once per month',
    'About once per week or more'
]

# convert purchase_freq to categorical with the specified order
df_clean['purchase_freq'] = pd.Categorical(
    df_clean['purchase_freq'], 
    categories=freq_order, 
    ordered=True
)

# Group by purchase type and disposal method, count the number of responses, and reshape for plotting
plot_df = df_clean.groupby(['purchase_freq', 'disposal_method_clean']).size().unstack('disposal_method_clean').fillna(0)

# Normalize by shopper type to get proportion of shopper groups that use each disposal method
sub_df = plot_df.div(plot_df.sum(axis=1), axis=0)

# Generate plot
ax = sub_df.plot(kind='barh', 
                 stacked=True, 
                 rot=0,
                 cmap='Set3',
                 title='Disposal Methods by Purchase Frequency')

# Customize plot labels
ax.legend(title='Disposal Method', loc='center left', bbox_to_anchor=(1.0, 0.5))
ax.set_xlabel("Proportion within group")
ax.set_ylabel("Purchase Frequency")

plt.show()
```

```python
# Let's also look at summary statistics to confirm the trends we see in the plot
df_clean.groupby('purchase_freq')['disposal_method_clean'].describe()
```

The stacked barplot above shows the proportion of disposal methods represented in each purchase frequency group. The plot shows:
- The most common disposal method for most purchase frequency groups is donation.
- The most common disposal method for the most frequent shoppers (once per week or more) is selling. 
- The purchase frequency group most likely to throw away clothes is the least frequent shoppers. 
- The purchase frequency group least likely to throw away clothes is the most frequent shoppers.

The last two points directly contradict our hypothesis that higher purchasing frequency is correlated with more throwing away of clothes.

Let's see if these relationships are statistically significant using a Chi-Square test.

```python
# Contingency table of raw counts for purchase type vs disposal method
contingency_table = pd.crosstab(df_clean['purchase_freq'], df_clean['disposal_method_clean'])

# Run the Chi-Square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Print the results
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
```

At a significance level of 5%, there is a statistically significant relationship between purchase frequency and disposal method because the p-value is 0.001. This means we fail to reject the null that purchase frequency and disposal method are independent.

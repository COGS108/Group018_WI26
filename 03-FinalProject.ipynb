## Background and Prior Work

Fast fashion is a clothing production model where clothing is produced quickly and cheaply, with rapid turnover due to changing trends and high-volume manufacturing. This method of production has significantly altered the fashion industry, influencing how trends spread and how consumers purchase clothing, particularly over the past two decades.

On one hand, this model has been praised for making clothing more affordable to consumers who may not otherwise be able to purchase expensive or high-quality garments. Lower prices allow a wider range of people to participate in fashion trends and access a greater variety of clothing styles. However, this accessibility also raises concerns about textile lifespan and sustainability. Garments are increasingly manufactured in ways that shorten their usable life, contributing to higher rates of waste, pollution, and resource consumption.
Prior research suggests that fast fashion production practices may directly reduce textile lifespan. According to the Sustainability Directory, fast fashion brands more often use inexpensive synthetic fibers and lower-quality materials in their clothing. These fabrics tend to degrade more quickly with repeated wear and washing, causing garments to lose structural integrity after a few uses.<a name="#ref1"></a>[<sup>1</sup>](#note1) While lower production costs make clothing more accessible, this tradeoff between affordability and durability may encourage consumers to replace garments more frequently, increasing the total volume of clothing discarded.

Consumer behavior also plays a major role in this cycle. The rapid release of new styles encourages consumers to replace clothing that is still wearable but no longer “in style.” According to the United States Environmental Protection Agency, the total mass of textile waste generated each year has increased roughly tenfold since 1960. <a name="#ref4"></a>[<sup>4</sup>](#note4) This increase suggests that changes in consumption patterns, rather than population growth alone, are a major contributor to textile waste.

International organizations report similar trends. The Geneva Environment Network estimates that approximately 85% of textiles are discarded each year, with the majority ending up in landfills or incinerators rather than being recycled.<a name="#ref2"></a>[<sup>2</sup>](#note2) Their findings indicate that the modern fashion industry prioritizes high production volume and low costs, but does not maintain clothing within long-term circulation systems such as repair, reuse, or recycling. As a result, garments often move quickly from production to disposal.

However, researchers have also identified challenges in accurately measuring the environmental impact of fast fashion. A Vox investigation by Alden Wicker highlights that many widely cited statistics about clothing waste originate from limited regional studies or outdated data.<a name="#ref3"></a>[<sup>3</sup>](#note3) While there is strong consensus that fast fashion contributes to overproduction and environmental strain, the exact scale of the problem remains difficult to quantify. Estimates of garment lifespan and disposal rates vary significantly between studies, and many rely on aggregated industry data rather than direct consumer behavior measurements. This uncertainty highlights an important limitation in existing research, where much of the available data focuses on large-scale industry production rather than the everyday clothing habits of individual consumers. 

Understanding how people actually purchase, wear, and dispose of clothing is essential for evaluating how fast fashion operates at the consumer level.
The lack of reliable consumer-level data surrounding the garment industry led us to examine clothing use and disposal practices within a specific population, in this case college students. College students represent a particularly relevant demographic because they are frequent consumers of fast fashion brands and often face financial constraints that influence purchasing decisions. By analyzing self-reported behaviors from this population, this study aims to provide more concrete insight into how clothing is purchased, used, and ultimately discarded. Our sources therefore provide a foundation for investigating how broader fast fashion trends may manifest in the everyday consumption patterns of college students.

### Tidying

```python
import pandas as pd
df = pd.read_csv('data/00-raw/Clothing Purchasing Behavior.csv')
df.columns = [
    'timestamp',
    'consent',
    'major',
    'new_vs_secondhand',
    'purchase_freq',
    'avg_price',
    'disposal_method',
    'disposal_reason'
]
# Give common disposal reasons their own columns; collect singletons in a column called 'other'
df_tidy = df.copy()
disposal_col_names = {
    "Doesn't fit anymore": "disposal_reason_doesnt_fit",
    "Don't like it anymore": "disposal_reason_dont_like",
    "Freeing up space": "disposal_reason_space",
    "Torn/Broken": "disposal_reason_torn",
    "Selling for money": "disposal_reason_money"
}
for reason, new_col in disposal_col_names.items():
    df_tidy[new_col] = df_tidy['disposal_reason'].str.contains(reason).astype(bool)
def is_other(response: str) -> bool:
    reasons = response.split(', ')
    for reason in reasons:
        if reason not in disposal_col_names:
            return True
    return False

df_tidy['disposal_reason_other'] = df_tidy['disposal_reason'].apply(is_other)
df_tidy = df_tidy.drop(columns=['disposal_reason'])

df_clean = df_tidy.copy()
# Major is the only column with missing values. Since we can make valid inferences without it, we will simply fill them in.
df_clean['major'] = df_clean['major'].fillna('Unknown')

# Standardize the text in the major column
df_clean['major'] = df_clean['major'].str.strip().str.lower()

# Respondent 3 had trouble selecting the survey options, so their response is invalid. Otherwise, everything looks OK.
df_clean = df_clean.drop(index=3)

# Drop nonconsenting participants.
df_clean = df_clean[df_clean['consent'] == "Yes"]
df_clean = df_clean.drop(columns=['consent'])

reason_cols = [
    "disposal_reason_doesnt_fit",
    "disposal_reason_dont_like",
    "disposal_reason_space",
    "disposal_reason_torn",
    "disposal_reason_money",
    "disposal_reason_other"
]
df_long = df_clean.melt(
    id_vars=["disposal_method"],
    value_vars=reason_cols,
    var_name="disposal_reason",
    value_name="selected"
)
df_long = df_long[df_long["selected"]]
reason_label_map = {
    "disposal_reason_doesnt_fit": "Doesn't fit",
    "disposal_reason_dont_like": "Don't like it",
    "disposal_reason_space": "Freeing space",
    "disposal_reason_torn": "Torn/Broken",
    "disposal_reason_money": "Selling for money",
    "disposal_reason_other": "Other"
}
df_long["disposal_reason"] = df_long["disposal_reason"].map(reason_label_map)

# map relevant "other" responses to the appropriate categories in disposal_method_clean
def simplify_method(method):
    method = str(method).lower()

    if "donate" in method or "thrift" in method or "hand down" in method:
        return "Donate"
    elif "sell" in method or "ebay" in method or "yard sale" in method:
        return "Sell"
    elif ("recycl" in method or "cut" in method or "scrap" in method or "patch" in method or "sewing" in method or "rag" in method or "cleaning" in method or "repurpose" in method):
        return "Recycle"
    elif "trash" in method or "throw" in method or "landfill" in method:
        return "Throw Away"
    elif "keep" in method or "closet" in method:
        return "Keep"
    else:
        return "Other"
df_long["disposal_method_clean"] = df_long["disposal_method"].apply(simplify_method)

# add disposal_method_clean to df_clean as well 
df_clean["disposal_method_clean"] = df_clean["disposal_method"].apply(simplify_method)
# drop the original disposal_method column since we have the cleaned version now
df_clean = df_clean.drop(columns=['disposal_method'])

df_clean.to_csv('data/02-processed/survey_clean2.csv', index=False)
df_clean.head()
```

#### Major Proportions: Does our survey sample reflect the general UCSD population by major?

Our goal with this EDA is to explore whether our sample is representative of the general UCSD population in terms of major, in the case that major may be  confounding varibale.

We considered that certain majors may be more likely to shop secondhand, such as environmental studies or fashion. However, we want to first see how "good" our data matches UCSD's distribution.

We will do this with a Chi-Square Goodness of Fit Test.

- Null Hypothesis: The survey sample proportion by major does NOT significantly differ from the expected proportions (from the UCSD degrees-conferred dataset). Hidden assumption: The UCSD degrees-conferred dataset is approximately equal to the number of people in a major
- Alternate Hypothesis: The survey sample proportion by major DOES significantly differ from the expected proportions

```python
# Keyword-Based Classifier Function: Maps Major keywords to 7 Categories
# This code block was generated with GPT-4 and edited to reflect common major keywords of UCSD students
def categorize_major(input_major):
    
    if any(word in input_major for word in [
        'literature', 'history', 'philosophy', 'art', 'music',
        'theatre', 'visual', 'foreign language', 'linguistics',
        'humanities', 'arts', 'theater', 'dance', 'english'
    ]):
        return 'humanities' # Arts & Humanities
    
    elif any(word in input_major for word in [
        'bio', 'neuro', 'chem', 'biological',
        'human biology', 'life science'
    ]):
        return 'bio' # Biological Sciences
   
    elif any(word in input_major for word in [
        'health', 'public health'
    ]):
        return 'health' # Health Sciences
    
    elif any(word in input_major for word in [
        'engineering', 'computer', 'electrical',
        'mechanical', 'structural', 'nano', 'aerospace', 'civil', 'industrial'
    ]):
        return 'engineering' # Engineering
    
    elif any(word in input_major for word in [
        'physical', 'physics', 'math', 'mathematics', 'statistics',
        'astronomy', 'physical science'
    ]):
        return 'physical' # Physical Sciences
    
    elif any(word in input_major for word in [
        'social', 'cognitive', 'psychology', 'econ', 'economics', 'sociology',
        'political', 'international', 'anthropology',
        'communication', 'business', 'finance'
    ]):
        return 'social' # Social Sciences
    
    else:
        return 'other'
```

```python
# Apply categorization method to df
df['category'] = df['major'].apply(categorize_major)
```

#### Shopper Type vs. Disposal Method

Let's investigate our hypothesis that shoppers who buy new or more likely to throw away unwanted clothes, while shoppers who buy secondhand are more likely to donate unwanted clothes. 

Let's look at the distribution of shopper types.

```python
# first mapping new_vs_secondhand into simpler categories for better visualization
new_vs_secondhand_map = {
   'Almost always new': 'New',
   'Mostly new, some secondhand': 'New',
   'About equal new and secondhand': 'Equal New & Secondhand',
   'Mostly secondhand, some new': 'Secondhand',
   'Almost always secondhand': 'Secondhand'
}

df['new_vs_secondhand'] = df['new_vs_secondhand'].map(new_vs_secondhand_map)

pie_chart = plt.pie(
    df['new_vs_secondhand'].value_counts(),
    labels=df['new_vs_secondhand'].value_counts().index
)

plt.title('Distribution of Shopper Types')
plt.axis('equal') 

plt.show()
```

Let's look at the distribution of disposal methods by shopper type.

```python
# Group by purchase type and disposal method, count the number of responses, and reshape for plotting
plot_df = df.groupby(['new_vs_secondhand', 'disposal_method_clean']).size().unstack('disposal_method_clean').fillna(0)

# Normalize by shopper type to get proportion of shopper groups that use each disposal method
sub_df = plot_df.div(plot_df.sum(axis=1), axis=0)

# Generate plot
ax = sub_df.plot(kind='bar', stacked=True, rot=0,
                      title='Disposal Methods by Shopper Type')

# Customize plot labels
ax.legend(title='Disposal Method', loc='center left', bbox_to_anchor=(1.0, 0.5))
ax.set_ylabel("Proportion within group")
ax.set_xlabel("Shopper Type")

plt.show()
```

The stacked barplot displays the proportion of disposal methods within each group of shoppers. It shows that the proportion of secondhand shoppers who donate their clothes is less than that of new shoppers. This contradicts our hypothesis that there is a higher proportion of donaters among secondhand shoppers compared to new shoppers. However, the plot does show a higher proportion of throwing away clothes among new shoppers compared to secondhand shoppers. Secondhand shoppers have the highest proportion of recyclers and sellers of unwanted clothes. As for shoppers who buy equal new and secondhand, they are the group with the highest proportion of donaters. 

Now, let's see if the relationship between shopper type and disposal method is statistically significant using a Chi-Square test. This is the appropriate metric because shopper type and disposal method are categorical variables, and disposal method is  nominal rather than ordinal. 

```python
# Contingency table of raw counts for purchase type vs disposal method
contingency_table = pd.crosstab(df['new_vs_secondhand'], df['disposal_method_clean'])

# Run the Chi-Square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Print the results
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
```

The p-value is 0.08, so, at the 5% significance level, the relationship between shopper type and disposal method is not statistically signifcant.

The last two points directly contradict our hypothesis that higher purchasing frequency is correlated with more throwing away of clothes.

Let's see if these relationships are statistically significant using a Chi-Square test. This is the appropriate metric because purchase frequency and disposal method are categorical variables, and disposal method is  nominal rather than ordinal. 

```python
# Contingency table of raw counts for purchase type vs disposal method
contingency_table = pd.crosstab(df_clean['purchase_freq'], df_clean['disposal_method_clean'])

# Run the Chi-Square test
chi2, p, dof, expected = chi2_contingency(contingency_table)

# Print the results
print(f"Chi-Square Statistic: {chi2}")
print(f"P-value: {p}")
print(f"Degrees of Freedom: {dof}")
```

At a significance level of 5%, there is a statistically significant relationship between purchase frequency and disposal method because the p-value is 0.001. This means we reject the null that purchase frequency and disposal method are independent.

Let's see the strength of the correlation between disposal method and purchase frequency using a Cramér's V test. We use this test because both categories are categorical.

```python
# This code was generated by Gemini 3

# 1. Total number of observations (n)
n = contingency_table.sum().sum()

# 2. Find the minimum dimension of the table (rows - 1 or cols - 1)
# Cramér's V uses the 'k' factor to account for table size
min_dim = min(contingency_table.shape) - 1

# 3. Calculate Cramér's V
v = np.sqrt(chi2 / (n * min_dim))

print(f"Cramér's V: {v:.4f}")
```

Given 0.10 < v < 0.30, there is a weak association between purchase frequency and disposal method.

- The shopper type most likely to dispose of clothes to free up space is secondhand shoppers.
- The shopper type most likely to dispose of clothes due to damage is new shoppers. 

Let's see if these relationships are statistically significant with a Chi-Square test for multiple responses. This is the appropriate metric because while both variables are categorical, a standard Chi-Square test could not account for the fact that respondents can select up to two disposal reasons. So, we run six separate Chi-Square tests and adjust the significant p-value accordingly. 

```python
# This code was generated by Gemini 3

results = []
alpha = 0.05
# Adjust alpha for multiple comparisons
bonferroni_alpha = alpha / len(reason_cols)

for reason in reason_cols:
    # 1. Create the table (Shopper Type vs. This Reason)
    contingency_table = pd.crosstab(df['new_vs_secondhand'], df[reason])
    
    # 2. Run Chi-Square
    chi2, p, dof, ex = chi2_contingency(contingency_table)
    
    results.append({
        'Disposal Reason': reason, 
        'P-Value': round(p, 4), 
        'Significant (Raw)': p < alpha,
        'Significant (Bonferroni)': p < bonferroni_alpha
    })

sig_df = pd.DataFrame(results)
print(sig_df)
```

#### Purchase Frequency vs. Disposal Reason

Our hypothesis suggests that fast fashion involves planned obsolescence—the clothes are made of poor quality so consumers need to re-purchase frequently. But is this really what drives purchasing frequency? Let's visualize the relationship between purchase frequency and disposal reason.

```python
# This code block was partially generated by Gemini 3

# Melt the boolean columns
data_long = df.melt(id_vars='purchase_freq', 
                  value_vars=reason_cols, 
                  var_name='disposal_reason', 
                  value_name='is_selected')

# Map reasons to more readable labels
reason_label_map = {
    'disposal_reason_doesnt_fit': "Doesn't fit",
    'disposal_reason_dont_like': "Don't like it",
    'disposal_reason_space': "Freeing up space",
    'disposal_reason_torn': "Torn/Broken",
    'disposal_reason_money': "Selling for money",
    'disposal_reason_other': "Other"
}
data_long['disposal_reason'] = data_long['disposal_reason'].map(reason_label_map)

# order purchase frequency for better visualization
freq_order = [
    'Every few years',
    'About once a year',
    'Every few months',
    'About once per month',
    'About once per week or more'
]

# convert purchase_freq to categorical with the specified order
df['purchase_freq'] = pd.Categorical(
    df['purchase_freq'], 
    categories=freq_order, 
    ordered=True
)

# Plotting (using the 'disposal_reason' column we just created)
ax = sns.barplot(x='purchase_freq', 
                 y='is_selected', 
                 hue='disposal_reason', 
                 data=data_long, 
                 estimator=lambda x: sum(x) / len(x) * 100)

sns.despine()
ax.set_title('Disposal Reasons by Purchase Frequency')
ax.set_ylabel('Percentage of Respondents in Group (%)')
ax.set_xlabel('Purchase Frequency')

# Rotate x-axis labels if they overlap
plt.xticks(rotation=45, ha='right')

ax.legend(title='Disposal Reason Selected (Up to 2)', loc='center left', bbox_to_anchor=(1.0, 0.5))

plt.show()
```

The grouped barplot above illustrates the distribution of disposal reasons within each purchase frequency group. Important patterns emerge:
- The group with the highest proportion of disposing due to damage is the least frequent shoppers (every few years)
- The group with the lowest proportion of disposing due to damage is the most frequent shoppers (about once per week or more)
- The most frequent shoppers mostly dispose of clothes due to not liking them anymore


These initial results contradict our assumption that shorter clothing lifespan is the reason for higher purchase frequency. This makes sense given our earlier finding that there is a correlation between lower purchasing frequency and throwing away clothes. 

Let's see if these relationships are statistically significant with a Chi-Square test for multiple responses. Again, this is the appropriate metric because while both variables are categorical, a standard Chi-Square test could not account for the fact that respondents can select up to two disposal reasons. We also use a Cramér's V test to see the strength of association between the two categorical variables. 

```python
# This code was generated by Gemini 3

results = []
alpha = 0.05
bonferroni_alpha = alpha / len(reason_cols)

for reason in reason_cols:
    # 1. Create the table
    contingency_table = pd.crosstab(df['purchase_freq'], df[reason])
    
    # 2. Run Chi-Square
    chi2, p, dof, ex = chi2_contingency(contingency_table)
    
    # 3. Calculate Cramér's V
    n = contingency_table.sum().sum()
    # k is the minimum of (number of rows - 1) or (number of columns - 1)
    min_dim = min(contingency_table.shape) - 1
    
    # Avoid division by zero for empty columns/rows
    if n > 0 and min_dim > 0:
        v = np.sqrt(chi2 / (n * min_dim))
    else:
        v = 0
    
    results.append({
        'Disposal Reason': reason, 
        'P-Value': round(p, 4), 
        'Cramér\'s V': round(v, 4),
        'Significant (Raw)': p < alpha,
        'Significant (Bonferroni)': p < bonferroni_alpha
    })

# Convert to DataFrame and sort by the strongest relationship
sig_df = pd.DataFrame(results).sort_values(by='Cramér\'s V', ascending=False)

print(sig_df)
```

Dislike and damage are the only disposal reasons with a statistically significant relationship to purchase frequency. Dislike has a weak association with purchase frequency, and Torn/Broken has a better, but still weak association with purchase frequency.

#### Shopper Type vs. Purchase Frequency

Our hypothesis assumes that buying new means buying more frequently because new clothes are often fast fashion and fast fashion has shorter release cycles, poorer quality, and perhaps a consumer base more interested in keeping up with trends. Let's visualize the relationship between shopper type and purchase frequency. This time, we will explore shopper type by calculating the proportion of clothes bought secondhand.

```python
secondhand_prop_map = {
    'Almost always new': 0.0,
    'Mostly new, some secondhand': 0.25,
    'About equal new and secondhand': 0.5,
    'Mostly secondhand, some new': 0.75,
    'Almost always secondhand': 1.0
}
df_clean['secondhand_prop'] = df_clean['new_vs_secondhand'].map(secondhand_prop_map)

purchase_freq_order = [
    'Every few years',
    'About once a year',
    'Every few months',
    'About once per month',
    'About once per week',
    'About once per week or more'
]

present_purchase_freq_order = [
    freq for freq in purchase_freq_order
    if freq in df_clean['purchase_freq'].dropna().unique()
]

ax = sns.pointplot(
    data=df_clean,
    x='purchase_freq',
    y='secondhand_prop',
    order=present_purchase_freq_order,
    estimator='mean',
    errorbar=('ci', 95),
    capsize=0.1
)
ax.yaxis.set_major_formatter(PercentFormatter(1.0))
ax.set_title('Purchase Frequency vs. Proportion Purchased Secondhand')
ax.set_xlabel('Purchase Frequency')
ax.set_ylabel('Secondhand Purchases (%)')
ax.tick_params(axis='x', rotation=20)

```

The pointplot shows that frequent shoppers have a lot of variation in the type of clothes they buy, ranging from 30 to 80% secondhand. However, the most frequent shoppers are by far the group that buys secondhand the most. This contradicts our assumption that buying new means buying more frequently.

Let's see if the observed pattern is statistically significant using a Kruskal-Wallis test. This is the appropriate metric since we are calculating shopper type as a numerical variable. We opt for a Kruskal-Wallis test rather than a one-way ANOVA because variance is not equal across purchase frequency groups. 

```python
# This code block was generated with Gemini 3
from scipy import stats

# 1. Prepare the data (ensure no NaNs)
df_nonparam = df_clean.dropna(subset=['purchase_freq', 'new_vs_secondhand'])

# 2. Group the data for the test
# Create a list of 'secondhand_prop' values for each 'purchase_freq' group
groups = [
    df_nonparam[df_nonparam['purchase_freq'] == freq]['new_vs_secondhand']
    for freq in present_purchase_freq_order
]

# 3. Perform the Kruskal-Wallis H-test
h_stat, p_val = stats.kruskal(*groups)

print("--- Kruskal-Wallis Result ---")
print(f"H-statistic: {h_stat:.4f}")
print(f"p-value: {p_val:.4e}")
```

There is no need to perform a post-hoc test because the p-value is too high. We fail to reject the null that there is no relationship between purchase frequency and proportion purchased secondhand.

3. Higher purchase frequency correlates with the disposal reason being damage.
